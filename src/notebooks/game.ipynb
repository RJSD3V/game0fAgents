```python
import numpy as N
import pygame 
import sys

def surfdemo_show(array_data, title="Array Preview"):


    pygame.init()

    width, height = array_data.shape[0], array_data.shape[1]
    screen = pygame.display.set_mode((width, height ))
    pygame.display.set_caption(title)


    if len(array_data.shape) == 2:

        surf = pygame.surfarray.make_surface(array_data)
    else:
        surf = pygame.surfarray.make_surface(array_data)

    screen.blit(surf, (0,0))
    pygame.display.flip()

    print(f"Showing {title} window. Close the Pygame window to continue ...")
    running = True
    while running: 
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
    pygame.quit()

```

```python

allblack = N.zeros((128, 128))
surfdemo_show(allblack, 'allblack')
```

```python
import numpy as np
import time
import random

class Grid:
    def __init__(self, size):
        self.size = size
        self.grid = [['0'] * self.size[0] for _ in range(self.size[1])]
        

    def create_cell(self):
        """
        Generates a 2D List array as a grid where the cells  will be alive or dead. 
        """
        posx = np.random.randint(0, self.size[0])
        print(f"Creating cell at position {posx}")
        posy = np.random.randint(0, self.size[1])
        print(f"Creating a cell at position {posx}, {posy}")

        self.grid[posx][posy] = '*'

    def show_grid(self):
        print(len(self.grid))
        for i in self.grid:
            for j in i:
                 time.sleep(0.0001)
                 print(f'{j}', end=' ')

            time.sleep(0.0002)
            print('\t')
            
```

```python
g = Grid((10, 10))
g.show_grid()
time.sleep(2)
g.create_cell()
g.show_grid()
g.create_cell()
g.show_grid()
g.create_cell()
g.show_grid()
```

```python
# This function will be vectorized so that it can be broadcasted across all elements. 
def moore_nb(x,y,b):
    # Sequence goes top,a nd moves anti-clockwise
    n_count = 0
    nbrs  = [(x-1,y), (x-1,y-1),(x,y-1), (x+1,y-1),(x+1,y),(x+1,y+1),(x,y+1),(x-1,y+1)]
    for (ix,iy) in nbrs:
        if 0<=ix < b.shape[0] and 0<= iy < b.shape[1]:
            if b[ix][iy] == 1:
                n_count +=1

    return n_count
        
```

```python
moore_nb_vec = np.vectorize(moore_nb, excluded=[2])
```

```python
neighbor_matrix = np.fromfunction(lambda x,y: moore_nb_vec(x,y,board),shape=(10, 10), dtype=int)
```

```python
# Filters

birth_condition = (neighbor_matrix == 3)
survival_mask = (neighbor_matrix == 2) | (neighbor_matrix == 3)
next_board = np.zeros_like(board)
next_board[(board == 0) & birth_condition] = 1
next_board[(board==1) & survival_mask] == 1
board = next_board
```
